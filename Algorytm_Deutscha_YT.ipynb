## 1. Wyrocznia

Implementacja funkcji generującej macierz 2x2
* **Input**: tuple z wartościami funkcji $f(0)$ i $f(1)$: `(f0, f1)`.
* Jeżeli użytkownik nie poda `(f0, f1)`, to wybieramy te wartości losowo.
* **Output**: Macierz 2x2, `np.array` z elementami $$U_f =  \begin{pmatrix} (-1)^{f(0)} & 0 \\ 0 & (-1)^{f(1)} \end{pmatrix}.$$


```python
def get_oracle(f = None):
    """
    Generates the oracle matrix.

    Input:
    - tuple with two values: f(0) and f(1)
      If not provided, the values are chosen randomly.

    Output:
    - 2x2 np.array containing the oracle matrix.
    """
    if f is None:
        f0 = np.random.randint(2)
        f1 = np.random.randint(2)
    else:
        f0, f1 = f

    print(f"f(0) = {f0}, f(1) = {f1}")
    if f0 == f1:
        print("The function is constant.")
    else:
        print("The function is balanced.")

    return np.diag([(-1)**f0, (-1)**f1])
```

# 2. Implementacja algorytmu Deutscha

Źródła:
* QNode: https://pennylane.ai/qml/glossary/quantum_node
* Device `"default.qubit"`: https://docs.pennylane.ai/en/stable/code/api/pennylane.device.html

### 2.1. Import/zainstaluj PennyLane

```python
# Setup the device
dev = qml.device("default.qubit", wires = 1, shots = 1)

# Define the quantum circuit function
@qml.qnode(dev)
def deutsch_quantum_circit(oracle):
    """
    Builds the Deutsch circuit and returns the sample of measurment outcome.

    Parameters:
    - oracle: 2x2 np.array containing the oracle matrix.

    Output:
    - measurement result.
    """
    qml.Hadamard(wires = 0)
    qml.QubitUnitary(oracle, wires = 0)
    qml.Hadamard(wires = 0)

    return qml.sample(qml.PauliZ(0))

# Define the function running the algorithm and printing the result
def run_deutsch_algorithm(oracle):
    """
    Performs Deutsch algorith and prints if the function is constant or balanced.

    Parameters:
    - oracle: 2x2 np.array containing the oracle matrix.

    Output:
    - None.
    """
    result = deutsch_quantum_circit(oracle)

    if result == 1:
        print("The function is constant.")
    elif result == -1:
        print("The function is balanced.")
    else:
        print("Something went wrong.")

    return result
```

## 3. Przeprowadzenie algorytmu
